```python
numeric_features = df_dedup[features].select_dtypes(include="number").columns

outlier_summary = []

for col in numeric_features:
    Q1 = df_dedup[col].quantile(0.25)
    Q3 = df_dedup[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df_dedup[col] < lower) | (df_dedup[col] > upper)).sum()

    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": outliers,
        "Outlier_Percentage": (outliers / len(df_dedup)) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    "Outlier_Percentage",
    ascending=False
)
```

### SMD (Standardized Mean Difference) Ranking

```python
import numpy as np
import pandas as pd

# Calculate mean and standard deviation for each class
grouped = df_dedup.groupby("Class")[features].agg(["mean", "std"])

# Extract statistics
mean_0 = grouped.loc[0].xs("mean", level=1)
mean_1 = grouped.loc[1].xs("mean", level=1)

std_0 = grouped.loc[0].xs("std", level=1)
std_1 = grouped.loc[1].xs("std", level=1)

# Standardized Mean Difference
smd = (mean_1 - mean_0) / np.sqrt(
    (std_1**2 + std_0**2) / 2
)

feature_effects = pd.DataFrame({
    "Mean_Legitimate": mean_0,
    "Mean_Fraud": mean_1,
    "SMD": smd,
    "Abs_SMD": abs(smd)
}).sort_values("Abs_SMD", ascending=False)

feature_effects
```

```python
from scipy.stats import ks_2samp

# ============================================
# CONSISTENT FEATURE SET
# ============================================

features = [f"V{i}" for i in range(1, 29)] + ["Time", "Amount"]


# ============================================
# 1. MEAN DIFFERENCE
# ============================================

summary_clean = df_dedup.groupby("Class")[features].mean().T

summary_clean.columns = ["Legitimate", "Fraud"]

summary_clean["Absolute Mean Difference"] = (
    summary_clean["Fraud"] - summary_clean["Legitimate"]
).abs()

mean_diff_rank = (
    summary_clean["Absolute Mean Difference"]
    .rank(ascending=False, method="min")
    .rename("Mean Difference Rank")
)


# ============================================
# 2. KS STATISTIC
# ============================================

ks_results_clean = []

for feature in features:

    legit = df_dedup.loc[
        df_dedup["Class"] == 0, feature
    ]

    fraud = df_dedup.loc[
        df_dedup["Class"] == 1, feature
    ]

    statistic, p_value = ks_2samp(legit, fraud)

    ks_results_clean.append({
        "Feature": feature,
        "KS Statistic": statistic,
        "p-value": p_value
    })

ks_results_clean = pd.DataFrame(ks_results_clean)

ks_rank = (
    ks_results_clean
    .set_index("Feature")["KS Statistic"]
    .rank(ascending=False, method="min")
    .rename("KS Rank")
)


# ============================================
# 3. SMD
# ============================================

grouped_clean = (
    df_dedup
    .groupby("Class")[features]
    .agg(["mean", "std"])
)

mean_0 = grouped_clean.loc[0].xs("mean", level=1)
mean_1 = grouped_clean.loc[1].xs("mean", level=1)

std_0 = grouped_clean.loc[0].xs("std", level=1)
std_1 = grouped_clean.loc[1].xs("std", level=1)

smd_clean = (
    (mean_1 - mean_0)
    / np.sqrt((std_1**2 + std_0**2) / 2)
)

smd_rank = (
    smd_clean.abs()
    .rank(ascending=False, method="min")
    .rename("SMD Rank")
)


# ============================================
# 4. PEARSON CORRELATION
# ============================================

corr_matrix_clean = (
    df_dedup[features + ["Class"]]
    .corr()
)

target_corr_clean = (
    corr_matrix_clean["Class"]
    .drop("Class")
)

pearson_rank = (
    target_corr_clean.abs()
    .rank(ascending=False, method="min")
    .rename("Pearson Rank")
)


# ============================================
# 5. COMBINE ALL FOUR RANKINGS
# ============================================

combined_ranking_clean = pd.concat(
    [
        mean_diff_rank,
        ks_rank,
        smd_rank,
        pearson_rank
    ],
    axis=1
)


# ============================================
# 6. AVERAGE RANK
# ============================================

combined_ranking_clean["Average Rank"] = (
    combined_ranking_clean[
        [
            "Mean Difference Rank",
            "KS Rank",
            "SMD Rank",
            "Pearson Rank"
        ]
    ].mean(axis=1)
)


# ============================================
# 7. FINAL CONSENSUS RANKING
# ============================================

combined_ranking_clean = (
    combined_ranking_clean
    .sort_values("Average Rank")
)

combined_ranking_clean
```

### Zero-Amount Transactions

There are 1,808 zero-amount transactions and no negative transaction amounts. Although zero-amount transactions represent a small portion of the dataset, their fraud rate is approximately 1.38%, compared with an overall fraud rate of approximately 0.17%. This indicates that zero-amount transactions may contain useful predictive information. Therefore, they are retained rather than treated as invalid observations.



### Outlier and Fraud Overlap

The IQR-based analysis shows that observations classified as outliers have a substantially higher fraud rate than non-outliers. Approximately 0.33% of outlier observations are fraudulent compared with approximately 0.01% of non-outlier observations.

This indicates that extreme observations may contain important fraud-related signal. Therefore, statistical outliers are retained rather than removed during preprocessing.
